```python
# Convert column names to snake_case
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(' ', '_')
    )

df.columns
```

```python
# checking null values

df.isnull().sum()
```

# The 'item_weight' column contained missing values that could impact downstream analysis. Since product weight is a fixed attribute for each product, missing values were first imputed using the average weight of the corresponding 'item_identifier'. Any remaining missing values were then filled using the median weight of the respective 'item_type' to maintain realistic product characteristics while minimizing bias.

```python
df['item_weight'] = df.groupby('item_identifier')['item_weight'].transform(
    lambda x: x.fillna(x.mean())
)

df.isnull().sum()
```

```python
# remaining null value in 'item_weight' is filled by the median weight of corresponding 'item_type'

df['item_weight'] = df.groupby('item_type')['item_weight'].transform(
    lambda x: x.fillna(x.median())
)

df['item_weight'].isnull().sum()
```

# checking unique values in 'item_fat_content' columns to identify inconsistent name conventions and spelling variations.

```python
df["item_fat_content"].unique()
```

# replacing different name of the same category

```python
df["item_fat_content"] = df["item_fat_content"].replace({
    "LF":"Low Fat",
    "low fat":"Low Fat",
    "reg":"Regular"
    })

# checking again

print(df["item_fat_content"].unique())
```

# investigating products with zero visibility values.

```python
df[df["item_visibility"] == 0][["item_identifier", "item_type", "sales", "item_visibility"]].head()
```

# products with zero visibility were found to have recorded sales,
# suggesting that zero values likely represent missing or incorectly recorded data. 
# these values were replaced with median visibility of products within the same item category

```python
df["item_visibility"] = df.groupby("item_type")["item_visibility"].transform(
    lambda x:x.replace(0,x[x > 0].median())
)
```

# creating sales category based on sales distribution for easier segementation and dashboard filtering

```python
df["sales_category"] = pd.qcut(df["sales"],q=3, labels=["Low", "Medium", "High"])

# checking

df.head()

```

# creating outlet_age to analyze whether older outlets perform differently from newer outlets.

```python
current_year = 2026

df["outlet_age"] = current_year - df["outlet_establishment_year"]

#checking

df.head()
```
